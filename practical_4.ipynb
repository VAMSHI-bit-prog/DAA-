```python

import time

# Iterative Factorial
# Time Complexity: O(n)
# Space Complexity: O(1)

def factorial_iterative(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result


def main():
    n = int(input("Enter a non-negative integer: "))

    if n < 0:
        print("Invalid input! Enter a non-negative integer.")
        return

    start = time.perf_counter_ns()
    result = factorial_iterative(n)
    end = time.perf_counter_ns()

    print("\n--- Iterative Factorial ---")
    print("Factorial Result :", result)
    print("Time Complexity  : O(n)")
    print("Space Complexity : O(1)")
    print("Execution Time   :", end - start, "nanoseconds")


if __name__ == "__main__":
    main()
```

The iterative factorial program successfully calculates the factorial of a given non-negative integer using a loop. It has a time complexity of O(n) and a space complexity of O(1), making it memory-efficient. The execution time is measured using Python's time.perf_counter_ns(). Therefore, the iterative method is simple, efficient, and uses constant extra space compared to the recursive method.

```python

import time

# Recursive Factorial
# Time Complexity: O(n)
# Space Complexity: O(n)

def factorial_recursive(n):
    if n <= 1:
        return 1

    return n * factorial_recursive(n - 1)


def main():
    n = int(input("Enter a non-negative integer: "))

    if n < 0:
        print("Invalid input! Enter a non-negative integer.")
        return

    start = time.perf_counter_ns()
    result = factorial_recursive(n)
    end = time.perf_counter_ns()

    print("\n--- Recursive Factorial ---")
    print("Factorial Result :", result)
    print("Time Complexity  : O(n)")
    print("Space Complexity : O(n)")
    print("Execution Time   :", end - start, "nanoseconds")


if __name__ == "__main__":
    main()
```

The recursive factorial program successfully calculates the factorial of a given non-negative integer by calling the function repeatedly until the base condition is reached. It has a time complexity of O(n) and a space complexity of O(n) due to the recursive call stack. The execution time is measured using Python's time.perf_counter_ns(). Thus, the recursive method is easy to understand but uses more memory than the iterative method.
