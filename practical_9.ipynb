```python
import time
import heapq

def prim(graph, n):
    visited = [False] * n
    min_heap = [(0, 0)]  # (weight, vertex)
    total_cost = 0
    mst_edges = []

    while min_heap:
        weight, u = heapq.heappop(min_heap)

        if visited[u]:
            continue

        visited[u] = True
        total_cost += weight

        if weight != 0:
            mst_edges.append((parent[u], u, weight))

        for v, w in graph[u]:
            if not visited[v]:
                heapq.heappush(min_heap, (w, v))
                parent[v] = u

    return total_cost, mst_edges


# ---------------- USER INPUT ----------------

n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

graph = [[] for _ in range(n)]
parent = [-1] * n

print("Enter each edge as: source destination weight")

for _ in range(e):
    u, v, w = map(int, input().split())

    # Convert to 0-based indexing
    u -= 1
    v -= 1

    graph[u].append((v, w))
    graph[v].append((u, w))  # Undirected graph


# ---------------- EXECUTION TIME ----------------

start_time = time.perf_counter()

total_cost, mst_edges = prim(graph, n)

end_time = time.perf_counter()

execution_time = end_time - start_time


# ---------------- OUTPUT ----------------

print("\nMinimum Spanning Tree:")

for u, v, w in mst_edges:
    print(f"{u + 1} -- {v + 1}  Weight = {w}")

print("\nTotal cost of MST:", total_cost)
print(f"Execution time: {execution_time:.10f} seconds")


# ---------------- COMPLEXITY ----------------

print("\nTime Complexity: O(E log V)")
print("Space Complexity: O(V + E)")
```

Prim's algorithm successfully finds the Minimum Spanning Tree (MST) of a connected, weighted, undirected graph. It uses a greedy approach to select the minimum-weight edges while connecting all vertices without forming cycles. The program calculates and displays the MST edges, total cost, and execution time. Its time complexity is O(E log V) using a priority queue, and its space complexity is O(V + E). Therefore, Prim's algorithm is useful for finding the minimum-cost connection between vertices in a graph.
