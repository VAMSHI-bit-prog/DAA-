```python
import time

def matrix_chain_order(p):
    n = len(p) - 1

    if n <= 0:
        return 0

    # Create DP table
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # Calculate minimum multiplication cost
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i] * p[k + 1] * p[j + 1]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[0][n - 1]


def main():
    n = int(input("Enter number of matrices: "))

    if n <= 0:
        print("Number of matrices must be positive.")
        return

    p = []

    print("\nEnter dimensions of the matrices:")

    rows = int(input("Enter rows of matrix 1: "))
    cols = int(input("Enter columns of matrix 1: "))

    if rows <= 0 or cols <= 0:
        print("Dimensions must be positive integers.")
        return

    p.append(rows)
    p.append(cols)

    for i in range(1, n):
        rows = int(input(f"Enter rows of matrix {i + 1}: "))

        if rows != p[-1]:
            print("Invalid dimensions! Columns of the previous matrix "
                  "must equal rows of the next matrix.")
            return

        cols = int(input(f"Enter columns of matrix {i + 1}: "))

        if cols <= 0:
            print("Dimensions must be positive integers.")
            return

        p.append(cols)

    # Measure execution time
    start_time = time.perf_counter()

    minimum_operations = matrix_chain_order(p)

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    # Display results
    print("\n--- Matrix Chain Multiplication Results ---")
    print("Minimum number of scalar multiplications:",
          minimum_operations)
    print(f"Execution Time: {execution_time:.8f} seconds")
    print("Time Complexity: O(n^3)")
    print("Space Complexity: O(n^2)")


if __name__ == "__main__":
    main()
```

The Matrix Chain Multiplication algorithm uses dynamic programming to find the most efficient order for multiplying a sequence of matrices. It minimizes the total number of scalar multiplications without changing the order of the matrices. The algorithm has a time complexity of O(n³) and a space complexity of O(n²), where

n is the number of matrices. Therefore, it provides an efficient method for optimizing matrix multiplication order and reducing computational cost.
